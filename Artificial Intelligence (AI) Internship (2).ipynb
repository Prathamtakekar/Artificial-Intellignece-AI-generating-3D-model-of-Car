```python
import torch
from diffusers import StableDiffusionPipeline,DiffusionPipeline
import trimesh
import os
```

```python
os.environ['HF_HOME'] = '~/cache'
```

```python
def generate_3d_from_text(prompt, output_dir="output"):
    os.makedirs(output_dir, exist_ok=True)
    print(f"Generating 3D model for prompt: {prompt}")
    # Login to Hugging Face using your token
    !huggingface-cli login # login if not alredy logged in
    from huggingface_hub import notebook_login
    notebook_login() # This will prompt you to enter your Hugging Face token.
    pipe = DiffusionPipeline.from_pretrained("OpenAI/point-e", torch_dtype=torch.float16)
    device = "cuda" if torch.cuda.is_available() else "cpu"
    pipe.to(device)
    # Moved the following lines inside the function
    pc = pipe(prompt).images[0] # Changed .image to .images
    output_path = os.path.join(output_dir, "model.obj")
    pc.save(output_path)
    print(f"3D model saved to: {output_path}")
    return output_path
```

```python
def visualize_3d_model(model_path):
  mesh = trimesh.load(model_path)
  mesh.show()
```

```python
prompt = "a small red toy car"
out_path = generate_3d_from_text(prompt)
visualize_3d_model(out_path)
```
